# EDA — Fetal Health Classification
Exploração do dataset de Cardiotocografia (CTG) para classificação de saúde fetal.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')

df = pd.read_csv('../data/raw/fetal_health.csv')
df = df.rename(columns={'baseline value': 'baseline_value'})
label_map = {1: 'Normal', 2: 'Suspect', 3: 'Pathological'}
df['label'] = df['fetal_health'].map(label_map)
print(df.shape)
df.head()
```

```python
# Distribuição das classes
fig, ax = plt.subplots(figsize=(6, 4))
colors = ['#2ecc71', '#f39c12', '#e74c3c']
df['label'].value_counts().plot(kind='bar', ax=ax, color=colors)
ax.set_title('Distribuição das Classes')
ax.set_xlabel('')
ax.set_ylabel('Contagem')
plt.tight_layout()
plt.show()
print(df['label'].value_counts(normalize=True).map('{:.1%}'.format))
```

```python
# Estatísticas descritivas
df.drop(columns=['fetal_health', 'label']).describe().T.style.background_gradient(cmap='Blues', subset=['mean', 'std'])
```

```python
# Boxplots das principais features por classe
features = ['baseline_value', 'accelerations', 'abnormal_short_term_variability',
            'mean_value_of_long_term_variability', 'histogram_mean']

fig, axes = plt.subplots(1, len(features), figsize=(18, 4))
palette = {'Normal': '#2ecc71', 'Suspect': '#f39c12', 'Pathological': '#e74c3c'}

for ax, feat in zip(axes, features):
    sns.boxplot(data=df, x='label', y=feat, palette=palette, ax=ax)
    ax.set_title(feat.replace('_', ' '), fontsize=9)
    ax.set_xlabel('')

plt.suptitle('Distribuição das Features por Classe', fontsize=12, y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Correlação com o target
numeric_df = df.drop(columns=['label'])
corr = numeric_df.corr()['fetal_health'].drop('fetal_health').sort_values(key=abs, ascending=False)

fig, ax = plt.subplots(figsize=(8, 6))
corr.plot(kind='barh', ax=ax, color=corr.map(lambda x: '#e74c3c' if x > 0 else '#2ecc71'))
ax.set_title('Correlação das Features com fetal_health')
ax.axvline(0, color='black', linewidth=0.8)
plt.tight_layout()
plt.show()
```

```python
# Heatmap de correlação (top 10 features)
top_feats = corr.abs().nlargest(10).index.tolist() + ['fetal_health']
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(numeric_df[top_feats].corr(), annot=True, fmt='.2f', cmap='RdBu_r', center=0, ax=ax)
ax.set_title('Correlação — Top 10 Features')
plt.tight_layout()
plt.show()
```
